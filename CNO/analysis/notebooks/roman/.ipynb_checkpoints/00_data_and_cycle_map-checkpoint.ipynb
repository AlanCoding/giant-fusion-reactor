# Roman workbook 00 — data and cycle map

This workbook establishes the input boundary for the five-recipe Roman baseline. It does **not** calculate chamber radii, driver allocations, ignition, or cycle closure. Its job is to make the working topology, units, and packaged nuclear datasets easy to inspect before any new physical model is selected. The baseline is conservative and reviewable; it is not yet claimed to be globally optimal.

```python
import cno_sweep
import pandas as pd
from IPython.display import display

from cno_sweep import BUILTIN_DATASETS, load_roman_mainline
from cno_sweep.constants import AVOGADRO
from cno_sweep.workbook import (
    neutron_cross_section_rows,
    reaction_rows,
    reactivity_rows,
    roman_chamber_rows,
)

pd.set_option('display.max_colwidth', 100)
print(f'cno_sweep {cno_sweep.__version__}')
display(pd.DataFrame(
    [{'dataset_id': key, 'packaged_resource': value} for key, value in BUILTIN_DATASETS.items()]
).sort_values('dataset_id', ignore_index=True))
```

## Accepted architecture

Every row below is one distinct central target recipe or shot chemistry; it does not require a permanently dedicated facility. Trajan is recorded separately as the distributed p+N15 driver and terminal catalyst-return reaction.

```python
manifest = load_roman_mainline()
display(pd.DataFrame(roman_chamber_rows()).sort_values('order', ignore_index=True))
display(pd.DataFrame([manifest['distributed_driver']]))
display(pd.DataFrame({'baseline_rule': manifest['baseline_rules']}))
assert manifest['central_chamber_count'] == 5
assert all(len(chamber['central_reaction_ids']) == 1 for chamber in manifest['chambers'])
assert manifest['chambers'][1]['excluded_from_central_fuel'] == ['h1']
```

## Mainline reaction ledger

The table exposes Q values, decay times, and automatic baryon/charge checks. It does not imply that the desired branch dominates competing channels.

```python
mainline_ids = [
    'c12-p-g-n13', 'n13-beta-c13',
    'c13-a-n-o16', 'n-p-g-d',
    'o16-p-g-f17', 'f17-beta-o17',
    'o17-p-a-n14',
    'n14-p-g-o15', 'o15-beta-n15',
    'n15-p-a-c12',
]
ledger = reaction_rows(mainline_ids)
display(pd.DataFrame(ledger))
assert all(row['baryon_conserved'] and row['charge_conserved'] for row in ledger)
```

## Rate comparison at T9 = 5

REACLIB reports molar rates in cm³ mol⁻¹ s⁻¹. The library evaluates particle-pair reactivity in m³/s, so the final expression converts back only for comparison with the architecture document. These are screening fits whose reaction-specific validity ranges still require review. Here `T9 = 5` is only a common nuclear-rate benchmark (`kT` about 431 keV), not an accepted self-supporting Trajan operating temperature.

```python
temperature_t9_5_keV = 5.0e9 / 1.160451812e7
rate_ids = [
    'c13-a-n-o16',
    'o16-p-g-f17',
    'o17-p-a-n14',
    'n14-p-g-o15',
    'n15-p-a-c12',
]
rates = reactivity_rows(rate_ids, [temperature_t9_5_keV])
for row in rates:
    row['na_reactivity_cm3_mol_s'] = row['reactivity_m3_s'] * AVOGADRO * 1.0e6

quoted_refactor_rates = {
    'c13-a-n-o16': 8.6237e5,
    'o16-p-g-f17': 1.22e3,
    'n15-p-a-c12': 9.57e7,
}
for row in rates:
    quoted = quoted_refactor_rates.get(row['reaction_id'])
    row['roman_refactor_quote_cm3_mol_s'] = quoted
    row['evaluated_over_quote'] = row['na_reactivity_cm3_mol_s'] / quoted if quoted else None

print(f'T9=5 corresponds to kT={temperature_t9_5_keV:.3f} keV')
display(pd.DataFrame(rates))
```

### Refactor-boundary discrepancy

With the pinned 2026-06-09 REACLIB subset, C13(alpha,n) evaluates to approximately **3.50e6 cm³ mol⁻¹ s⁻¹ at T9=5**, while `ROMAN_REFACTOR.md` quotes approximately **8.62e5**. O16(p,gamma) and N15(p,alpha) reproduce the document's quoted values. Workbook `05_nuclear_data.ipynb` must compare the exact fit sets and determine whether they are additive contributions or alternative evaluations. This workbook intentionally displays the packaged value rather than silently forcing agreement.

## Constantine neutron-data inventory

This is only a lookup from the currently packaged **light-nuclide ENDF/B-VIII.0 subset** at two screening energies. Workbooks 05 and 32 must extend the versioned data boundary with C13(p,gamma) and Pb-isotope data respectively. Workbooks 40 and 45 must add the source spectrum, time-dependent composition, scattering, inverse reactions, complete radial geometry, expansion, and a recoverable-D criterion.

```python
constantine_xs = neutron_cross_section_rows(
    ['h1', 'he4', 'c13', 'o16', 'n14', 'n15'],
    [2.0, 14.1],
)
constantine_xs_table = pd.DataFrame(constantine_xs)
constantine_xs_table['nonelastic_barn'] = (
    constantine_xs_table['total_barn'] - constantine_xs_table['elastic_barn']
)
constantine_xs_table['other_nonelastic_barn'] = (
    constantine_xs_table['nonelastic_barn'] - constantine_xs_table['capture_barn']
)
display(constantine_xs_table)
```

## Review checklist and boundary carried forward

This workbook asks the reviewer to confirm only the following boundary:

1. Five separate central target recipes are the conservative starting topology, not a proven optimum or five necessarily separate facilities.
2. Trajan is the distributed N15+p driver and terminal catalyst-return reaction, not a sixth central recipe.
3. The packaged C13(alpha,n) rate discrepancy remains open and blocks trusted Constantine dimensions.
4. The current neutron table is an inventory of available data, not a transport result.
5. No radius, compression, pusher allocation, D gain, or reference-design viability is claimed here.

A later workbook must not assume that a neutron which deposits energy becomes D, or that D created in hot fuel survives. Constantine succeeds only if its neutron leaves the hot source, is captured on suitably placed ordinary hydrogen, and produces D that survives into a recoverable inventory.
