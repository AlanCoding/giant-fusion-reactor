# Workbook 40 — common reaction and target-radius envelopes

This workbook applies **one calculation flow** to all five Roman central-fuel recipes. Its immediate deliverable is the likely/conservative estimate of:

- initial reaction-fuel radius $R_0$;
- compressed fuel radius $R_f$;
- p+N15/DT Trajan-driver thickness;
- Pb-tamper thickness;
- complete target radius, masses, burnup, and yield.

These are **Phase-I workbook estimates** intended to supply parameters to later simulations. They are not simulation-qualified target designs. In particular, the calculation first asks how large a uniformly hot compressed sphere must be to burn before disassembly. It then reports how fast a real central burn front would have to travel, because the fixed DT kernel cannot heat these enormous cores uniformly.

```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import Markdown, display

repo_root = Path.cwd()
while repo_root.name != 'CNO' and repo_root != repo_root.parent:
    repo_root = repo_root.parent
source_root = repo_root / 'analysis' / 'src'
if str(source_root) not in sys.path:
    sys.path.insert(0, str(source_root))

from cno_sweep.layered_driver import evolve_layered_pressure_pulse
from cno_sweep.n15_pusher import additive_volume_density
from cno_sweep.reaction_envelope import (
    ROMAN_REACTION_RECIPES, DriverBracket, layered_target_estimate,
    minimax_n15_budget_selection, reaction_radius_state,
)

pd.set_option('display.max_columns', 30)
pd.set_option('display.precision', 5)
print('Common radius model loaded for', ', '.join(ROMAN_REACTION_RECIPES))
```

## Radius logic

For one heavy nucleus $A$ and an initial partner ratio $r=n_B/n_A$, both reactants deplete. The exact fixed-state time to burn fraction $f$ is

$$t_f=\frac{f}{(1-f)n_{A0}\langle\sigma v\rangle} \qquad (r=1),$$

and, when $r\ne1$,

$$t_f=\frac{\ln[(r-f)/(r(1-f))]}{(r-1)n_{A0}\langle\sigma v\rangle}.$$

The effective whole-fuel disassembly time is

$$\tau_h=C_h\frac{R_f}{c_s}.$$

The marginal reaction-race radius follows by setting $t_f=\tau_h$:

$$R_f=\frac{c_s t_f}{C_h},\qquad R_0=R_f C^{1/3}.$$

Density comes from a volume-additive condensed-mixture estimate followed by compression $C=\rho_f/\rho_0$. The sound speed uses ideal ions plus the finite-temperature Fermi-electron pressure. This is a uniform-state necessary-condition model. It does not calculate shocks, radial temperature gradients, reaction-front propagation, or instability.

## Driver and Pb logic

For each candidate compressed state, the library calculates reversible cold-electron compression work for the reaction fuel and fixed central DT kernel. A declared mechanical fraction $\eta_{cold}$ converts that useful work into required deposited Trajan/DT driver energy:

$$E_{driver,dep}=\frac{W_{cold,fuel}+W_{cold,DT}}{\eta_{cold}}.$$

Loaded N15 is sized from its burned 4.966-MeV yield plus the declared DT-vein contribution. Driver mass then fixes its volume and thickness. Pb mass is four times active-driver mass and fixes the outer shell thickness.

The five targets are coupled by the steady catalyst constraint

$$\sum_i \frac{N_{15, burned,i}}{N_{desired,successful,i}}=1.$$

The discrete minimax search first enforces this as a maximum: it selects the smallest possible **largest complete target radius** on the stated grids with a calculated driver demand no greater than one. Any residual below one is not spare catalyst; it must be added back as harmless driver overdrive in one or more targets so that the final ledger is exactly one. This is why no recipe receives an unlimited private N15 pusher. The scalar $\eta_{cold}$ is calibrated only as a likely/conservative bracket from workbook 35; a pressure-chamber audit at the end shows where that shortcut misses.

```python
recipe_rows = [{
    'recipe': r.name, 'reaction id': r.reaction_id,
    'fuel': f'{r.heavy_nuclide} + {r.partner_nuclide}',
    'Q (MeV)': r.q_mev, 'neutron role': r.neutron_role,
} for r in ROMAN_REACTION_RECIPES.values()]
pd.DataFrame(recipe_rows).set_index('recipe')
```

## Likely and conservative search branches

These are editable engineering brackets, not fitted confidence intervals. The likely branch permits higher temperature, up to four protons per heavy ion in proton reactions, 50% desired burn, and a full $R_f/c_s$ global clock. The conservative branch stops at 300 keV, permits at most two protons, asks for 80% burn, and uses half a sound-crossing time. Constantine remains exactly 1:1 C13:alpha and proton-free in both.

The central DT radii retain the previously discussed 8-cm likely and 25-cm conservative inventory choices. They are not claimed optimum. Driver DT:N15 ratios are inherited from the present vein-spacing screen and remain simulation inputs.

```python
case_inputs = {
    'likely': {
        'burn_fraction': 0.50, 'C_h': 1.0,
        'temperatures_keV': np.linspace(100.0, 430.0, 12),
        'compressions': np.geomspace(3e3, 2e6, 20),
        'proton_ratios': (1.0, 2.0, 4.0),
        'bracket': DriverBracket(
            name='likely', mechanical_cold_work_fraction=0.25,
            driver_proton_ratio=4.0, driver_density_kg_m3=additive_volume_density(4.0),
            n15_burn_fraction=0.50, dt_pairs_loaded_per_n15_loaded=0.0556,
            dt_burn_fraction=0.80, dt_neutron_deposition_fraction_in_driver=0.20,
            tamper_to_driver_mass_ratio=4.0, tamper_density_kg_m3=11340.0,
            central_dt_initial_radius_m=0.08, central_dt_initial_density_kg_m3=250.0,
            central_dt_burn_fraction=0.50, central_dt_neutron_deposition_fraction_in_core=0.50,
        ),
    },
    'conservative': {
        'burn_fraction': 0.80, 'C_h': 0.50,
        'temperatures_keV': np.linspace(80.0, 300.0, 12),
        'compressions': np.geomspace(2e3, 1e6, 20),
        'proton_ratios': (1.0, 2.0),
        'bracket': DriverBracket(
            name='conservative', mechanical_cold_work_fraction=0.12,
            driver_proton_ratio=4.0, driver_density_kg_m3=additive_volume_density(4.0),
            n15_burn_fraction=0.35, dt_pairs_loaded_per_n15_loaded=0.08,
            dt_burn_fraction=0.60, dt_neutron_deposition_fraction_in_driver=0.05,
            tamper_to_driver_mass_ratio=4.0, tamper_density_kg_m3=11340.0,
            central_dt_initial_radius_m=0.25, central_dt_initial_density_kg_m3=250.0,
            central_dt_burn_fraction=0.30, central_dt_neutron_deposition_fraction_in_core=0.20,
        ),
    },
}
case_rows = []
for name, case in case_inputs.items():
    b = case['bracket']
    case_rows.append({
        'case': name, 'desired burn': case['burn_fraction'], 'C_h': case['C_h'],
        'temperature range (keV)': f"{case['temperatures_keV'][0]:g}–{case['temperatures_keV'][-1]:g}",
        'compression range': f"{case['compressions'][0]:.2g}–{case['compressions'][-1]:.2g}",
        'proton ratios': str(case['proton_ratios']), 'cold-work fraction': b.mechanical_cold_work_fraction,
        'N15 burn fraction': b.n15_burn_fraction, 'DT pairs / loaded N15': b.dt_pairs_loaded_per_n15_loaded,
        'central DT R0 (m)': b.central_dt_initial_radius_m, 'Pb/driver mass': b.tamper_to_driver_mass_ratio,
    })
pd.DataFrame(case_rows).set_index('case')
```

## Nuclear-rate input

All rates come from the pinned JINA REACLIB snapshot packaged with `cno_sweep`. The source snapshot does not provide reaction-specific high-temperature validity intervals. The 430-keV endpoint is therefore a screening limit, not a validation claim.

```python
from cno_sweep import load_builtin_rate
temperature_plot = np.geomspace(30.0, 430.0, 140)
fig, ax = plt.subplots(figsize=(8.5, 5))
for recipe in ROMAN_REACTION_RECIPES.values():
    rate = load_builtin_rate(recipe.reaction_id)
    ax.loglog(temperature_plot, [rate.rate_m3_s(t) for t in temperature_plot], label=recipe.name)
ax.set(xlabel='ion temperature kT (keV)', ylabel=r'$\langle\sigma v\rangle$ (m$^3$/s)', title='Pinned reaction rates used by the radius screen')
ax.grid(True, which='both', alpha=0.3)
ax.legend()
plt.tight_layout()
```

## Shared-budget minimax search

The search evaluates each recipe on the declared temperature, compression, and partner-ratio grid. At every possible maximum target radius it chooses the least-N15-cost card available for each recipe below that radius. The first set whose five calculated driver requirements sum to at most one is the discrete minimax solution. The displayed deficit from exactly one is subsequently available only as driver overdrive; it is not a net N15 surplus.

```python
candidate_sets = {}
selected_sets = {}
selection_costs = {}
for case_name, case in case_inputs.items():
    candidates = {}
    for recipe_id, recipe in ROMAN_REACTION_RECIPES.items():
        ratios = (1.0,) if recipe_id == 'constantine' else case['proton_ratios']
        candidates[recipe_id] = []
        for ratio in ratios:
            for compression in case['compressions']:
                for temperature in case['temperatures_keV']:
                    state = reaction_radius_state(
                        recipe, partner_ratio=ratio, compression_ratio=float(compression),
                        ion_temperature_keV=float(temperature),
                        target_heavy_burn_fraction=case['burn_fraction'],
                        geometric_confinement_coefficient=case['C_h'],
                    )
                    candidates[recipe_id].append(layered_target_estimate(state, case['bracket']))
    selected, cost = minimax_n15_budget_selection(candidates, 1.0)
    candidate_sets[case_name] = candidates
    selected_sets[case_name] = selected
    selection_costs[case_name] = cost

pd.DataFrame([
    {'case': name, 'N15 burned / traversal': selection_costs[name],
     'largest complete target R0 (m)': max(x.physical_target_outer_radius_m for x in selected_sets[name].values())}
    for name in selected_sets
]).set_index('case')
```

## The requested radius table

`fuel R0` is the equivalent initial radius of reaction fuel alone. `physical core R0` adds the central DT-kernel volume. The Trajan mantle begins there; the Pb shell begins at `driver outer R0`; `complete R0` is the outside of Pb. Thickness is a physical radial difference, not an areal-density shorthand.

```python
geometry_rows = []
for case_name, selected in selected_sets.items():
    for recipe_id, target in selected.items():
        s = target.radius_state
        geometry_rows.append({
            'case': case_name, 'recipe': s.recipe.name,
            'fuel R0 (m)': s.initial_fuel_equivalent_radius_m,
            'fuel Rf (m)': s.compressed_fuel_radius_m,
            'physical core R0 incl DT (m)': target.physical_core_outer_radius_m,
            'Trajan+DT thickness (m)': target.driver_thickness_m,
            'driver outer R0 (m)': target.driver_outer_radius_m,
            'Pb thickness (m)': target.tamper_thickness_m,
            'complete R0 (m)': target.physical_target_outer_radius_m,
        })
geometry_table = pd.DataFrame(geometry_rows).set_index(['case', 'recipe'])
geometry_table
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4.8))
for ax, case_name in zip(axes, ('likely', 'conservative')):
    table = geometry_table.loc[case_name]
    x = np.arange(len(table))
    ax.plot(x, table['physical core R0 incl DT (m)'], 'o-', label='core outside')
    ax.plot(x, table['driver outer R0 (m)'], 'o-', label='Trajan outside')
    ax.plot(x, table['complete R0 (m)'], 'o-', label='Pb outside')
    ax.set_xticks(x, table.index, rotation=35, ha='right')
    ax.set(ylabel='initial radius (m)', title=case_name)
    ax.grid(True, alpha=0.3)
    ax.legend()
plt.tight_layout()
```

## Why each fuel radius has that value

This table exposes every quantity in the reaction-race equation. `rhoR` is extremely large because the slow reactions require a huge reaction column. Compression changes physical radius strongly but changes the marginal reaction column only weakly. Constantine's desired neutron recovery is therefore not demonstrated by its apparently modest physical radius; workbook 45 must transport that neutron through this column.

```python
burn_rows = []
for case_name, selected in selected_sets.items():
    for target in selected.values():
        s = target.radius_state
        burn_rows.append({
            'case': case_name, 'recipe': s.recipe.name, 'partner/heavy': s.partner_ratio,
            'desired burn': s.target_heavy_burn_fraction, 'compression': s.compression_ratio,
            'kT (keV)': s.ion_temperature_keV, 'rho_f (kg/m3)': s.compressed_density_kg_m3,
            '<sv> (m3/s)': s.reactivity_m3_s, 'burn time (s)': s.exact_burn_time_s,
            'hydro time (s)': s.hydrodynamic_time_s, 'sound speed (m/s)': s.sound_speed_m_s,
            'fuel Rf (m)': s.compressed_fuel_radius_m, 'fuel R0 (m)': s.initial_fuel_equivalent_radius_m,
            'rhoR (kg/m2)': s.rho_r_kg_m2, 'kT/E_F': s.temperature_over_fermi,
        })
burn_table = pd.DataFrame(burn_rows).set_index(['case', 'recipe'])
burn_table
```

## Mass, driver allocation, and yield

The Pb shell is radially thin because lead is dense, although its mass is fixed at four times the active driver. `N15 cost` is normalized to one successfully completed central reaction; its sum over the five selected rows is the shared cycle cost shown above. Fusion yield includes desired central reactions, burned Trajan N15, driver DT, and the fixed central DT kernel. It is not net electricity.

```python
mass_rows = []
for case_name, selected in selected_sets.items():
    for target in selected.values():
        s = target.radius_state
        mass_rows.append({
            'case': case_name, 'recipe': s.recipe.name,
            'fuel mass (kg)': s.fuel_mass_kg, 'central DT mass (kg)': target.central_dt_mass_kg,
            'Trajan+vein mass (kg)': target.driver_mass_kg, 'Pb mass (kg)': target.tamper_mass_kg,
            'total target mass (kg)': target.total_target_mass_kg,
            'N15 cost / successful reaction': target.n15_burned_per_successful_reaction,
            'driver DT burned / success': target.driver_dt_burned_per_successful_reaction,
            'central DT burned / success': target.central_dt_burned_per_successful_reaction,
            'total fusion yield (J)': target.total_fusion_yield_j,
        })
mass_table = pd.DataFrame(mass_rows).set_index(['case', 'recipe'])
mass_table
```

## The unresolved central-heating problem

The radius equation assumes the compressed fuel is hot. The architecture instead contains a finite central DT kernel. The table compares deposited energy from that kernel with the energy required to raise the entire compressed reaction fuel uniformly to the selected temperature.

`equivalent hot radius fraction` is the cube root of the affordable hot mass fraction. It is only an energy-equivalent central volume. `required front speed` is the remaining compressed radial distance divided by one calculated hydrodynamic time. It is the pass condition for the later spherical burn simulation, not a predicted speed.

```python
heating_rows = []
for case_name, selected in selected_sets.items():
    for target in selected.values():
        s = target.radius_state
        heating_rows.append({
            'case': case_name, 'recipe': s.recipe.name,
            'cold compression / heavy (MeV)': s.cold_compression_mev_per_initial_heavy,
            'uniform heat / heavy (MeV)': s.uniform_thermal_mev_per_initial_heavy,
            'desired fusion / heavy (MeV)': s.desired_fusion_mev_per_initial_heavy,
            'uniform core heat (J)': target.uniform_core_heating_energy_j,
            'central DT deposited (J)': target.central_dt_deposited_fusion_j,
            'DT / uniform heat': target.central_dt_uniform_heating_fraction,
            'equivalent hot radius fraction': target.equivalent_hot_radius_fraction,
            'required front speed (m/s)': target.required_core_front_speed_m_s,
        })
heating_table = pd.DataFrame(heating_rows).set_index(['case', 'recipe'])
heating_table
```

## Pressure-chamber consistency audit

The selected geometry used a scalar bracket for speed. As a check—not a correction—the existing two-moving-boundary pressure-chamber model is run with the resulting N15/DT loading per initial heavy unit. `achieved/selected C` should be near or above one. Large disagreement means the driver thickness is not yet self-consistent and must be iterated in workbook 35 before promotion. The reaction-race fuel radius itself is unchanged by this audit.

```python
mechanical_rows = []
for case_name, selected in selected_sets.items():
    b = case_inputs[case_name]['bracket']
    for recipe_id, target in selected.items():
        s = target.radius_state
        loaded_n15_per_heavy = target.loaded_n15_nuclei / s.initial_heavy_nuclei
        abundances = {s.recipe.heavy_nuclide: 1.0, s.recipe.partner_nuclide: s.partner_ratio}
        result = evolve_layered_pressure_pulse(
            event_id=f'audit-{case_name}-{recipe_id}', initial_core_radius_m=1.0,
            initial_core_density_kg_m3=s.initial_density_kg_m3, initial_abundances=abundances,
            driver_n15_loaded_per_core_unit=loaded_n15_per_heavy, driver_proton_ratio=b.driver_proton_ratio,
            n15_burn_fraction=b.n15_burn_fraction,
            dt_pairs_loaded_per_core_unit=b.dt_pairs_loaded_per_n15_loaded * loaded_n15_per_heavy,
            dt_burn_fraction=b.dt_burn_fraction, dt_neutron_deposition_fraction=b.dt_neutron_deposition_fraction_in_driver,
            driver_density_kg_m3=b.driver_density_kg_m3, tamper_to_driver_mass_ratio=b.tamper_to_driver_mass_ratio,
            tamper_density_kg_m3=b.tamper_density_kg_m3, burn_duration_over_characteristic_time=0.1,
            maximum_compression_ratio=1e8, trace_points=30,
        )
        mechanical_rows.append({
            'case': case_name, 'recipe': s.recipe.name, 'selected C': s.compression_ratio,
            'pressure-model C': result.peak_compression_ratio,
            'achieved/selected C': result.peak_compression_ratio / s.compression_ratio,
            'assumed cold-work fraction': b.mechanical_cold_work_fraction,
            'pressure-model cold-work fraction': result.useful_cold_compression_fraction,
            'status': 'passes selected C' if result.peak_compression_ratio >= s.compression_ratio else 'driver must be iterated',
        })
mechanical_audit = pd.DataFrame(mechanical_rows).set_index(['case', 'recipe'])
mechanical_audit
```

## First-pass interpretation

- **Diocletian and Aurelian consume most of the shared driver budget.** Their proton captures are the scale-setting slow reactions over the allowed temperature range.
- **Caesar remains important in physical radius** because its selected low-cost compression is modest, even though its N15 cost is small.
- **Constantine and Scipio are smaller by reaction kinetics**, but their low driver load makes the scalar mechanical fraction least reliable. Constantine additionally faces the separate neutron-escape/recovery problem.
- The fixed central DT kernel heats only a tiny energy-equivalent fraction of the largest selected cores. Consequently, the values here are useful simulation targets only if a several-megametre-per-second reaction/heat front or distributed heating mechanism exists.
- The enormous masses and yields are consequences of applying condensed initial density to tens- or hundreds-of-metres radii. They must feed the blast-chamber workbook without being normalized away.

## What these numbers do not prove

1. Uniform temperature is not an ignition mechanism. The central DT deposition and required-front-speed columns expose the missing spatial step.
2. The selected compression ratios are not yet achieved self-consistently for every row. Use the pressure-chamber audit rather than the scalar fraction where they disagree.
3. High-temperature REACLIB extrapolation and competing reaction channels remain nuclear-data uncertainties.
4. Bremsstrahlung, gamma transport, conduction, shocks, entropy, preheat, instability, and mix do not evolve in the radius equation.
5. Constantine neutron recovery is not implied by a successful C13 burn. Its enormous `rhoR` enters workbook 45.
6. Driver DT and central DT enter a preliminary D/T cost here, but full tritium makeup and neutron credits belong in the exact global ledger.
7. The minimax result is grid-dependent. The workbook is meant to identify the scale and critical inequalities before continuous optimization or spatial simulation.
